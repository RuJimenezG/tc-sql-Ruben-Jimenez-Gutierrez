## Conexión a BigQuery

```python
# Importar módulos
import os
from dotenv import load_dotenv
from google.cloud import bigquery
from google.oauth2 import service_account

# Cargar variables de entorno
load_dotenv()
PROJECT_ID = os.getenv("GCP_PROJECT_ID")
DATASET_ID = os.getenv("BQ_DATASET_ID")
CREDENTIALS_PATH = os.getenv("GOOGLE_APPLICATION_CREDENTIALS")

# Autenticación y cliente
credentials = service_account.Credentials.from_service_account_file(
    CREDENTIALS_PATH
)
client = bigquery.Client(
    project=PROJECT_ID,
    credentials=credentials
)

print(f"Conectado a BigQuery. Proyecto: {PROJECT_ID}")
```

## Crear el dataset en BigQuery

```python
# Crear dataset
dataset_ref = bigquery.Dataset(f"{PROJECT_ID}.{DATASET_ID}")
dataset_ref.location = "EU"  # Datos en Europa
dataset = client.create_dataset(dataset_ref, exists_ok=True)
print(f"Dataset {DATASET_ID} creado")
```

## Creación de tablas

El orden de creación es el siguiente:
  1. customers y categories
  2. orders y products
  3. payments y order_items
  4. reviews

```python
# Tabla customers

schema = [
    bigquery.SchemaField("customer_number", "INT64", mode="REQUIRED"),
    bigquery.SchemaField("first_name", "STRING", mode="REQUIRED"),
    bigquery.SchemaField("last_name", "STRING", mode="REQUIRED"),
    bigquery.SchemaField("email", "STRING", mode="REQUIRED"),
    bigquery.SchemaField("country", "STRING", mode="REQUIRED"),
    bigquery.SchemaField("city", "STRING", mode="REQUIRED"),
    bigquery.SchemaField("channel", "STRING", mode="REQUIRED"),
    bigquery.SchemaField("registration_date", "DATE", mode="REQUIRED"),
]

table_ref = f"{PROJECT_ID}.{DATASET_ID}.customers"
table = bigquery.Table(table_ref, schema=schema)
table = client.create_table(table, exists_ok=True)
print(f"Tabla customers creada")
```

```python
# Tabla categories

schema = [
    bigquery.SchemaField("id", "INT64", mode="REQUIRED"),
    bigquery.SchemaField("category_name", "STRING", mode="REQUIRED"),
    bigquery.SchemaField("description_name", "STRING", mode="NULLABLE"),    
]

table_ref = f"{PROJECT_ID}.{DATASET_ID}.categories"
table = bigquery.Table(table_ref, schema=schema)
table = client.create_table(table, exists_ok=True)
print(f"Tabla categories creada")
```

```python
# Tabla orders

schema = [
    bigquery.SchemaField("order_number", "INT64", mode="REQUIRED"),
    bigquery.SchemaField("customer_number", "INT64", mode="REQUIRED"),
    bigquery.SchemaField("order_status", "STRING", mode="REQUIRED"),
    bigquery.SchemaField("order_date", "TIMESTAMP", mode="REQUIRED"),
    bigquery.SchemaField("shipping_address", "STRING", mode="REQUIRED"),
    bigquery.SchemaField("shipping_date", "TIMESTAMP", mode="NULLABLE"),
    bigquery.SchemaField("delivery_date", "TIMESTAMP", mode="NULLABLE"),
]

table_ref = f"{PROJECT_ID}.{DATASET_ID}.orders"
table = bigquery.Table(table_ref, schema=schema)
table = client.create_table(table, exists_ok=True)
print(f"Tabla orders creada")
```

```python
# Tabla products

schema = [
    bigquery.SchemaField("product_number", "INT64", mode="REQUIRED"),
    bigquery.SchemaField("product_name", "STRING", mode="REQUIRED"),
    bigquery.SchemaField("category_id", "INT64", mode="REQUIRED"),
    bigquery.SchemaField("price", "NUMERIC", mode="REQUIRED"),
    bigquery.SchemaField("cost", "NUMERIC", mode="REQUIRED"),
    bigquery.SchemaField("stock", "STRING", mode="REQUIRED"),
    bigquery.SchemaField("is_active", "BOOL", mode="REQUIRED"),
]

table_ref = f"{PROJECT_ID}.{DATASET_ID}.products"
table = bigquery.Table(table_ref, schema=schema)
table = client.create_table(table, exists_ok=True)
print(f"Tabla products creada")
```

```python
# Tabla payments

schema = [
    bigquery.SchemaField("id", "INT64", mode="REQUIRED"),
    bigquery.SchemaField("order_number", "INT64", mode="REQUIRED"),
    bigquery.SchemaField("method", "STRING", mode="REQUIRED"),
    bigquery.SchemaField("payment_status", "STRING", mode="REQUIRED"),
    bigquery.SchemaField("amount", "NUMERIC", mode="REQUIRED"),
    bigquery.SchemaField("payment_date", "TIMESTAMP", mode="REQUIRED"),
]

table_ref = f"{PROJECT_ID}.{DATASET_ID}.payments"
table = bigquery.Table(table_ref, schema=schema)
table = client.create_table(table, exists_ok=True)
print(f"Tabla payments creada")
```

```python
# Tabla order_items

schema = [
    bigquery.SchemaField("id", "INT64", mode="REQUIRED"),
    bigquery.SchemaField("order_number", "INT64", mode="REQUIRED"),
    bigquery.SchemaField("product_number", "INT64", mode="REQUIRED"),
    bigquery.SchemaField("quantity", "INT64", mode="REQUIRED"),
    bigquery.SchemaField("buy_price", "NUMERIC", mode="REQUIRED"),
    bigquery.SchemaField("discount", "NUMERIC", mode="NULLABLE"),
]

table_ref = f"{PROJECT_ID}.{DATASET_ID}.order_items"
table = bigquery.Table(table_ref, schema=schema)
table = client.create_table(table, exists_ok=True)
print(f"Tabla order_items creada")
```

```python
# Tabla reviews

schema = [
    bigquery.SchemaField("id", "INT64", mode="REQUIRED"),
    bigquery.SchemaField("order_item_id", "INT64", mode="REQUIRED"),
    bigquery.SchemaField("rating", "INT64", mode="REQUIRED"),
    bigquery.SchemaField("comment", "STRING", mode="NULLABLE"),
]

table_ref = f"{PROJECT_ID}.{DATASET_ID}.reviews"
table = bigquery.Table(table_ref, schema=schema)
table = client.create_table(table, exists_ok=True)
print(f"Tabla reviews creada")
```
